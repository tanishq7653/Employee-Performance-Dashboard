```python
department_analysis = df.groupby('Department').agg(
    Employees=('Employee_ID', 'count'),
    Attritions=('Attrition_Flag', 'sum'),
    Avg_Salary=('Monthly_Salary', 'mean'),
    Avg_Performance=('Performance_Score', 'mean')
)

department_analysis['Attrition_Rate_%'] = (
    department_analysis['Attritions'] /
    department_analysis['Employees'] * 100
)

department_analysis.sort_values(
    'Attrition_Rate_%',
    ascending=False
)
```

```python
tenure_analysis = df.groupby(
    'Tenure_Band',
    observed=False
).agg(
    Employees=('Employee_ID', 'count'),
    Attritions=('Attrition_Flag', 'sum')
)

tenure_analysis['Attrition_Rate_%'] = (
    tenure_analysis['Attritions'] /
    tenure_analysis['Employees'] * 100
)

tenure_analysis['Attrition_Share_%'] = (
    tenure_analysis['Attritions'] /
    df['Attrition_Flag'].sum() * 100
)

tenure_analysis
```

```python
satisfaction_analysis = df.groupby(
    'Job_Satisfaction'
).agg(
    Employees=('Employee_ID', 'count'),
    Attritions=('Attrition_Flag', 'sum')
)

satisfaction_analysis['Attrition_Rate_%'] = (
    satisfaction_analysis['Attritions'] /
    satisfaction_analysis['Employees'] * 100
)

satisfaction_analysis
```

```python
satisfaction_analysis['Attrition_Rate_%'].plot(
    kind='bar',
    title='Attrition Rate by Job Satisfaction'
)

plt.ylabel('Attrition Rate (%)')
plt.show()
```

```python
wlb = df.groupby(
    'Work_Life_Balance'
).agg(
    Employees=('Employee_ID', 'count'),
    Attritions=('Attrition_Flag', 'sum')
)

wlb['Attrition_Rate_%'] = (
    wlb['Attritions'] /
    wlb['Employees'] * 100
)

wlb
```

```python
overtime = df.groupby('Overtime').agg(
    Employees=('Employee_ID', 'count'),
    Attritions=('Attrition_Flag', 'sum')
)

overtime['Attrition_Rate_%'] = (
    overtime['Attritions'] /
    overtime['Employees'] * 100
)

overtime
```
